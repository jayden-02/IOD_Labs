# *Data Cleaning with Pandas*

In this lab we work with a file that has messy column names and a combination of missing and invalid data to be cleaned.

To start with load `rock.csv` into a dataframe.

```python
# ANSWER
import pandas as pd
import numpy as np

from google.colab import files
uploaded = files.upload()
```

```python
rock_csv = 'rock.csv'
rock = pd.read_csv(rock_csv)
```

### Check Column Names

Check column names and clean by converting all names to lowercase, replacing spaces with underscores (`_`) and removing question marks or asterisk (`*`) characters.

```python
print(rock.columns)
```

```python
# ANSWER
rock.columns = rock.columns.str.lower()  # Convert all column names to lowercase
rock.columns = rock.columns.str.replace(' ', '_')  # Replace spaces with underscores
rock.columns = rock.columns.str.replace('[\?*]', '', regex=True)  # Remove question marks and asterisks

print(rock.columns)
```

### Replace Null Values With 0

Check 'release_year' column whether this column has any null values or not. Replace null values with 0.

```python
# ANSWER
print(rock['release_year'])
null_values = rock['release_year'].isnull().sum()
print(null_values)
rock['release_year'] = rock['release_year'].fillna(0)
null_values_fixed = rock['release_year'].isnull().sum()
print(null_values_fixed)
```

## Check Datatypes of Dataset

Check datatypes of the dataset. Is there any column which should be int instead of object? Fix the column.

```python
# ANSWER
print(rock.dtypes)
#released_year should be an int

rock['release_year'] = pd.to_numeric(rock['release_year'], errors='coerce', downcast='integer') #float
rock['release_year'] = rock['release_year'].fillna(0).astype(int) #float to int
print(rock.dtypes)
```

## Check Min, Max of Each Column

Is there any illogical value in any column? How can we fix that?

```python
# ANSWER
print("Min values:")
print(rock.min())

print("\nMax values:")
print(rock.max())

#Looking at the min/max values, only the release year is illogical. We could replace 0 with NaN.
rock['release_year'] = rock['release_year'].replace(0, np.nan)

print("Min values:")
print(rock.min())

print("\nMax values:")
print(rock.max())
```

### Write a function that will take a row of a DataFrame and print out the song, artist, and whether or not the release date is < 1970

```python
def print_song_info(row):
    song = row['song_clean']
    artist = row['artist_clean']
    print(f"Song: {song}")
    print(f"Artist: {artist}")
    if pd.isna(row['release_year']):
      print(f"Release Year: NaN")
    elif row['release_year'] < 1970:
      print(f"Release Year: before 1970")
    else:
      print(f"Release Year: after or in 1970")
    print("-" * 40)
(rock.iloc[:10]).apply(print_song_info, axis=1) #testing first 10 rows
(rock.iloc[-10:]).apply(print_song_info, axis=1) #testing last 10 rows
print(rock.tail()) #verifying with the tail. The song "She Loves my Automobile" has year==NaN, and the function correctly outputs Release Year: NaN
```

### Write a function that converts a column in a DataFrame to a numeric type and otherwise replaces entries with np.nan

```python
# ANSWER
def convert_to_numeric(df, column_name):
    df[column_name] = pd.to_numeric(df[column_name], errors='coerce') #float. errors='coerce' otherwise replaces entries with np.nan
    return df
```

### Apply this last function to your dataset

```python
# ANSWER
convert_to_numeric(rock, 'release_year')
```
