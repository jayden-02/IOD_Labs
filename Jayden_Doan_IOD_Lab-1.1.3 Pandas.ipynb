#### 2.D Describe `age`

Find the count, mean, std, min, max of the column 'age'

````python
# A:
count = df["age"].count()
mean = df["age"].mean()
std = df["age"].std()
min_age = df["age"].min()
max_age = df["age"].max()

print("Count:", count)
print("Mean:", mean)
print("Standard Deviation:", std)
print("Min:", min_age)
print("Max:", max_age)
````

# 4. Visualisation


#### 4.A Plot a histogram of `age`

````python
# A:
plt.hist(df["age"], bins=10, edgecolor="black")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
````

#### 4.B BoxPlot of `age`

````python
# A:
plt.boxplot(df["age"], vert=False, patch_artist=True)
plt.xlabel("Age")
plt.title("Box Plot of Age")
plt.show()
````

#### 4.C Create a scatterplot of `age` and `length_of_service`

````python
# A:
plt.scatter(df["age"], df["length_of_service"], alpha=0.5, color="blue")
plt.xlabel("Age")
plt.ylabel("Length of Service (Years)")
plt.title("Scatter Plot of Age vs. Length of Service")
plt.show()
````

#### 4.D Print a statistical summary of `age`

````python
# A:
print(df["age"].describe())
````
