# 9. [BONUS] Plot X, Y and Best Fit Line

**Write a function to calculate best fit.**

![alt text](https://i.imgur.com/B43vOV0.jpg)

[Least Squares Regression (Best Line Fit)]: https://www.mathsisfun.com/data/least-squares-regression.html "Least Squares Regression"

```
# Expected Output:
```

````python
"""CODE DELETED. DO NOT TOUCH THIS TEXT. RUNNING THIS CELL WITH SHIFT-ENTER WILL DELETE THE GRAPH"""
````

````python
# sample points
X = [0, 5, 10, 15, 20]
Y = [0, 7, 10, 13, 20]

# solve for a and b
def best_fit(X, Y):
    '''
    Return m, b
    # ANSWER
    '''

    m = (len(X)*sum(np.multiply(X,Y)) - sum(X)*sum(Y))/(len(X)*sum(np.multiply(X,X))-sum(X)**2)
    b = (sum(Y)-m*sum(X))/len(X)

    return m, b

# solution
m, b = best_fit(X, Y)

# plot points and fit line
# ANSWER
plt.scatter(X,Y)
plt.plot(X,m*np.array(X)+b)
plt.show()
print("best fit line:")
print(f"y = {b} + {m}x")
````

# 13. [BONUS] Calculate Mean Confidence Interval

Check the following methods:

1. scipy.stats.sem
2. scipy.stats.t.ppf

Expected output:
```
(172.52, 168.26329573448135, 176.77670426551867)
```


````python
#def mean_confidence_interval(data, confidence=0.95):
# Calculate m, se

def mean_confidence_interval(data, confidence=0.95):
  a = 1.0 * np.array(data)
  n = len(a)
  m, se = scipy.stats.tmean(a), scipy.stats.sem(a)
  h = se * scipy.stats.t.ppf((1 + confidence) / 2., n-1)
  return m, m-h, m+h

data = [170,167,	174,	179,	179,	187,	179,	183,	179 ,156,	163	,156	,187,	156	,
        167	,156	,174	,170 ,183	,179	,174	,179	,170,	159	,187	]
print(mean_confidence_interval(data))
````
